```python
data = [99,2,3,5,7,28,38,47]
total= sum(data)
count=len(data)
mean=total/count
minvalue=min(data)
maxvalue=max(data)
print(f'Data={data}')
print(f'Total={total}')
print(f'Count={count}')
print(f'Mean={mean}')
print(f'Min={minvalue}')
print(f'Max={maxvalue}')
```
